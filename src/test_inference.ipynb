# TEST INFERENCE PIPELINE

```python
from __future__ import annotations

from pathlib import Path
import numpy as np
import torch
from typing import Any

from emma import EmmaDataset, EmmaSearcher, SearchQuery
from torch.utils.data import DataLoader
from emma.dataloader import EmmaMapDataset
from transformers import AutoTokenizer, CLIPTextModelWithProjection

import utils.logging_ as logging
import time
```

```python
# Check CUDA Availability

if torch.cuda.is_available():
    print("Ready")
    device = torch.device('cuda')
else:
    print("Not Ready")
    device = torch.device('cpu')
```

## > DATASET

```python
dataset_root = None  # e.g. Path("/path/to/EMMa")

if dataset_root is None:
    raise ValueError("Please set `dataset_root` to the path of the EMMa dataset.")

dataset_root = Path(dataset_root)

if not dataset_root.exists():
    raise FileNotFoundError(f"Dataset path does not exist: {dataset_root}")
```

```python
# INIT DATASET
ds = EmmaDataset(dataset_root)
```

```python
# QUERY
se = EmmaSearcher(ds)
query = SearchQuery(splits=["Test"], require_listed=["mass", "category"], require_present=["package_dimensions"], quantity_range=[1,1], limit=10)
asins = se.query_asins(query)
print(len(asins))
```

```python
def make_transform(ds: EmmaDataset):
    def transform(asin: str, rec: dict) -> dict:
        return {
            "asin": asin,
            "title": ds.get_title(rec=rec),
            "categories": ds.get_category_names(rec=rec),
        }

    return transform


def collate_samples(samples: list[dict]) -> list[dict]:
    return samples
```

```python
loader = DataLoader(
    EmmaMapDataset(ds, asins, transform=make_transform(ds)),
    batch_size=4, shuffle=False, collate_fn=collate_samples,
)
```

## > MODEL

```python
clip_root = None  # e.g. Path("/path/to/CLIP")

if clip_root is None:
    raise ValueError("Please set `clip_root` to the path of the CLIP model.")

clip_root = Path(clip_root)

if not clip_root.exists():
    raise FileNotFoundError(f"Model path does not exist: {clip_root}")
```

```python
tokenizer = AutoTokenizer.from_pretrained(clip_root)
model = CLIPTextModelWithProjection.from_pretrained(
    clip_root,
    use_safetensors=True,
).to(device)
model.eval()
model.requires_grad_(False)
print(f"CLIP text encoder loaded on {device}")
```

## > INFERENCE

```python
def flatten_category_hierarchy(categories: Any) -> list[str]:
    if isinstance(categories, str):
        value = categories.strip()
        return [value]
    if isinstance(categories, (list, tuple)):
        hierarchy: list[str] = []
        for category in categories:
            hierarchy.extend(flatten_category_hierarchy(category))
        return hierarchy

    value = str(categories).strip()
    return [value]


def sample_text(sample: dict) -> str:
    hierarchy = flatten_category_hierarchy(sample["categories"])
    return " > ".join(hierarchy)


@torch.inference_mode()
def encode_texts(
    model: CLIPTextModelWithProjection,
    tokenizer: Any,
    device: torch.device,
    texts: list[str],
) -> torch.Tensor:
    inputs = tokenizer(
        texts,
        padding=True,
        truncation=True,
        max_length=model.config.max_position_embeddings,
        return_tensors="pt",
    ).to(device)

    embeddings = model(**inputs).text_embeds

    return embeddings.float().cpu()
```

```python
start = time.time()

for batch_idx, batch in enumerate(loader):
    texts = [sample_text(sample) for sample in batch]
    embeddings = encode_texts(model, tokenizer, device, texts).numpy()

    for sample, text, embedding in zip(batch, texts, embeddings):
        print("INPUT: ", text)

    logging.progress_bar(batch_idx, len(loader), 1, 1)

print(f"Elapsed: {time.time() - start:.2f}s")
```
